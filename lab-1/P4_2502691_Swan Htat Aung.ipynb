Firstly let's import the necessary libraries.

```python
### there are different ways to implement, this is one example
### 

from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy import Column, Integer, String, Enum, Float, ForeignKey, create_engine
from sqlalchemy.orm import relationship, declarative_base, sessionmaker

# Create SQLite engine, Base, and Session
engine = create_engine("sqlite:///students.db", echo=False)
Base = declarative_base()
Session = sessionmaker(bind=engine)
session = Session()
```


## Q1.
Define a database table as a class, and create the table in the database. The class name is ‘Students’, and the table name is ‘INF2003<your name>’, e.g., INF2003Tong. The table has 3 columns, namely sid (for student identifier as an integer and primary key), name (for the name of a student), and grade (for the grade of a student).

```python
# Q1: your codes go here, you can add more cells here if needed

class Students(Base):
    __tablename__ = "INF2003Aung"
    
    sid = Column(Integer, primary_key=True)
    name = Column(String)
    grade = Column(String)

    def __init__(self, sid, name, grade):
        self.sid = sid
        self.name = name
        self.grade = grade

# Create the table in the database
Base.metadata.create_all(engine)
```

##Q2.
Perform the 1st **read** operation in the table just created. Display all the rows, if any, in the table and check if the table is empty.

(Note: Every display operation in our tasks is a **read** operation.)


```python
# Q2: your codes go here, you can add more cells here if needed

students = session.query(Students).all()
if not students:
    print("The table is empty.")
else:
    for student in students:
        print(f"sid: {student.sid}, name: {student.name}, grade: {student.grade}")
```


##Q3.
Perform the 1st **insert** operation by adding yourself as a new student (with sid <your student id>, name <your name>, and grade ‘NA’) into the table.

Display all the rows in the latest table.


```python
# Q3: your codes go here, you can add more cells here if needed

my_student = Students(sid=2201234, name="Aung", grade="NA")
session.add(my_student)
session.commit()

students = session.query(Students).all()
for student in students:
    print(f"sid: {student.sid}, name: {student.name}, grade: {student.grade}")
```

##Q4.
Perform the 2nd and 3rd **insert** operations by adding a new student
(**sid 2202345, name ‘Arya Stark’, and grade ‘NA’**) and another new student (**sid 2203456, name ‘Jon Snow’, and grade ‘NA’**) into the table.

Display all the rows in the latest table.

```python
# Q4: your codes go here, you can add more cells here if needed

student_arya = Students(sid=2202345, name="Arya Stark", grade="NA")
student_jon = Students(sid=2203456, name="Jon Snow", grade="NA")

session.add_all([student_arya, student_jon])
session.commit()

students = session.query(Students).all()
for student in students:
    print(f"sid: {student.sid}, name: {student.name}, grade: {student.grade}")
```

##Q5.
Perform the 1st **update** operation by updating the grade of Arya Stark to ‘A-’. Display all the rows in the latest table.

(Note: A good practice to check if the item is in the table first.)

```python
# Q5: your codes go here, you can add more cells here if needed

arya = session.query(Students).filter_by(name="Arya Stark").first()
if arya:
    arya.grade = "A-"
    session.commit()

students = session.query(Students).all()
for student in students:
    print(f"sid: {student.sid}, name: {student.name}, grade: {student.grade}")
```

##Q6.
Perform the 2nd **update** operation by updating the grade of Jon Snow to ‘A+’.

Display all the rows in the latest table.

```python
# Q6: your codes go here, you can add more cells here if needed

jon = session.query(Students).filter_by(name="Jon Snow").first()
if jon:
    jon.grade = "A+"
    session.commit()

students = session.query(Students).all()
for student in students:
    print(f"sid: {student.sid}, name: {student.name}, grade: {student.grade}")
```

##Q7.
Perform the 1st **delete** operation by deleting the student of ‘Arya Stark’.

Display all the rows in the latest table.

```python
# Q7: your codes go here, you can add more cells here if needed

arya = session.query(Students).filter_by(name="Arya Stark").first()
if arya:
    session.delete(arya)
    session.commit()

students = session.query(Students).all()
for student in students:
    print(f"sid: {student.sid}, name: {student.name}, grade: {student.grade}")
```

##Q8.
Perform the 2nd **delete** operation by deleting all the students from the table.

Check and display what are the rows in the latest table.

```python
# Q8: your codes go here, you can add more cells here if needed

session.query(Students).delete()
session.commit()

students = session.query(Students).all()
if not students:
    print("The table is empty.")
else:
    for student in students:
        print(f"sid: {student.sid}, name: {student.name}, grade: {student.grade}")
```
