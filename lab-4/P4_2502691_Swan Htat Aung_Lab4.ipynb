##Q1.
With Python library **sqlite3**, create (with connect) a database called “music.sqlite” and delete the table called “Tracks” if it exists. Then, you create a new table called “**Tracks**” with 2 columns:

“title” as TEXT

“plays” as Integer

(You may think deleting Tracks in the beginning is unnecessary. Is this true?)

```python
#Q1 you codes go here, you can have more cells if needed

import sqlite3
conn = sqlite3.connect('music.sqlite')
cur = conn.cursor()
cur.execute('DROP TABLE IF EXISTS Tracks')
cur.execute('CREATE TABLE Tracks (title TEXT, plays INTEGER)')
conn.commit()
```

##Q8.
1) Create another database “order.db”.

2) Delete two tables “customers” and “orders” if any.

3) Then, create table “customers” with 2 columns, “customer_id” as Integer as primary key and “customer_name” as TEXT.

4) Create another table “orders” with 3 columns, “order_id” as TEXT as primary key, “customer_id” as Integer, and “amount” as DOUBLE.
Meanwhile, specify a foreign key as “customer_id” referring to the “customer_id” in table “customers”.


```python
#Q8 you codes go here, you can have more cells if needed

conn = sqlite3.connect('order.db')
cur = conn.cursor()
cur.execute("PRAGMA foreign_keys = ON")
cur.execute("DROP TABLE IF EXISTS orders")
cur.execute("DROP TABLE IF EXISTS customers")
cur.execute('CREATE TABLE customers (customer_id INTEGER PRIMARY KEY,customer_name TEXT)')
cur.execute('CREATE TABLE orders (order_id TEXT PRIMARY KEY,customer_id INTEGER,amount DOUBLE,FOREIGN KEY (customer_id) REFERENCES customers(customer_id))')
conn.commit()
```

##Q9.
Insert the below tuples into table “customers”. Print the latest table.

(1, 'Arya')

(2, 'Jon')

(3, 'Robb')

(4, '\<Yourself\>'), which is the name of yourself

##Q10.
Insert the below tuples into table “orders”.

 Print the latest table.

##Q11.
Join table “customers” and table “orders” together based on column “customer_id”.

Print the table after join with both tables’ data. (Do we have any information about Jon in the output?)

```python
#Q11 you codes go here, you can have more cells if needed

cur.execute('SELECT * FROM customers JOIN orders ON customers.customer_id = orders.customer_id')
for row in cur:
    print(row)
```

##Q12.
In the above question, you practiced how to return multiple columns from different tables. Here, return the “customer_name” and “order_id” of all the orders each with an amount greater than 15.

```python
#Q12 you codes go here, you can have more cells if needed

cur.execute('SELECT customers.customer_name, orders.order_id FROM customers JOIN orders ON customers.customer_id = orders.customer_id WHERE orders.amount > 15')
for row in cur:
    print(row)
```
